```python
import os
import soundfile as sf
import numpy as np
from datasets import load_dataset, Audio

TEST_DIR = "data/test"
os.makedirs(f"{TEST_DIR}/clean", exist_ok=True)
os.makedirs(f"{TEST_DIR}/noisy", exist_ok=True)

print("Loading dataset...")
dataset = load_dataset("MeiWu1123/VoiceBank-DEMAND-16k", split="test")
dataset = dataset.cast_column("clean", Audio(sampling_rate=16000))
dataset = dataset.cast_column("noisy", Audio(sampling_rate=16000))

print(f"Writing {len(dataset)} test samples to disk...")
for i, sample in enumerate(dataset):
    clean = np.array(sample["clean"]["array"], dtype=np.float32)
    noisy = np.array(sample["noisy"]["array"], dtype=np.float32)
    fname = f"{i:04d}.wav"
    sf.write(f"{TEST_DIR}/clean/{fname}", clean, 16000)
    sf.write(f"{TEST_DIR}/noisy/{fname}", noisy, 16000)

print("Done.")
```

```python
patch = """
import torch
_original_torch_load = torch.load
def _patched_torch_load(*args, **kwargs):
    kwargs['weights_only'] = False
    return _original_torch_load(*args, **kwargs)
torch.load = _patched_torch_load
"""

with open('/content/sgmse/enhancement.py', 'r') as f:
    content = f.read()

# inject after the first import torch line
content = content.replace('import torch\n', 'import torch\n' + patch, 1)

with open('/content/sgmse/enhancement.py', 'w') as f:
    f.write(content)

print("Patched successfully")
```
